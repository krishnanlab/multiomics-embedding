```python
import os
import pandas as pd

raw_feats = pd.read_csv('../data/raw/microbe_metabolites_filtered_rank_normalized.csv')
feats = list(raw_feats.columns)
feats.remove('sample')
met = [f for f in feats if 'N_AQ.' in f or 'N_LP.' in f]
micro = [f for f in feats if f not in met]

with open('../data/raw/metabolites.txt', 'w') as f:
    f.write('\n'.join(met))

with open('../data/raw/microbes.txt', 'w') as f:
    f.write('\n'.join(micro))

models = ['wcksnlsg','ai9n4jxs','7o4yga2v','21tdsqsa','q2gzu1o3','8lofhbbf','qb4y98x0']
out_root = '../results/best/zscores'
os.makedirs(out_root, exist_ok=True)
for model in models:
    for model_type in ['diet', 'time']:
        fp = f"../results/best/{model}_{model_type}_feature_predictions.tsv"
        preds = pd.read_csv(fp, sep='\t', index_col=0)
        zscored_met = (preds.loc[met] - preds.loc[met].mean()) / preds.loc[met].std()
        zscored_micro = (preds.loc[micro] - preds.loc[micro].mean()) / preds.loc[micro].std()
        zscored_met.to_csv(f"{out_root}/{model}_{model_type}_metabolites_predictions_zscored.tsv", sep='\t')
        zscored_micro.to_csv(f"{out_root}/{model}_{model_type}_microbes_predictions_zscored.tsv", sep='\t')
```

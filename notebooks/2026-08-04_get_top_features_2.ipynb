```python
import pandas as pd
with open("../data/nodes/microbes.txt", "r") as f:
    microbes = [line.strip() for line in f.readlines()]

with open("../data/nodes/metabolites.txt", "r") as f:
    metabolites = [line.strip() for line in f.readlines()]
```

```python
diet_perm = pd.read_csv("../results/permutations/diet_significant_mean_confidence.tsv", sep="\t")
diet_micro = diet_perm[diet_perm["feature"].isin(microbes)]
diet_metab = diet_perm[diet_perm["feature"].isin(metabolites)]
diet_micro, diet_metab
meat_metab = diet_metab[diet_metab["direction"] == "meat"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()
dairy_metab = diet_metab[diet_metab["direction"] == "dairy"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()

with open("../results/permutations/meat_metabolites_top_10.txt", "w") as f:
    for feature in meat_metab:
        f.write(feature + "\n")
with open("../results/permutations/dairy_metabolites_top_10.txt", "w") as f:
    for feature in dairy_metab:
        f.write(feature + "\n")
    
meat_micro = diet_micro[diet_micro["direction"] == "meat"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()
dairy_micro = diet_micro[diet_micro["direction"] == "dairy"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()
with open("../results/permutations/meat_microbial_top_10.txt", "w") as f:
    for feature in meat_micro:
        f.write(feature + "\n")
with open("../results/permutations/dairy_microbial_top_10.txt", "w") as f:
    for feature in dairy_micro:
        f.write(feature + "\n")
```

```python
diet_majority = pd.read_csv("../results/deployment_for_permutations/majority_hits_7.tsv", sep="\t")
diet_micro = diet_majority[diet_majority["feature"].isin(microbes)]
diet_metab = diet_majority[diet_majority["feature"].isin(metabolites)]
meat_micro = diet_micro[diet_micro["predicted_class"] == "diet_meat"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
dairy_micro = diet_micro[diet_micro["predicted_class"] == "diet_dairy"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
meat_metab = diet_metab[diet_metab["predicted_class"] == "diet_meat"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
dairy_metab = diet_metab[diet_metab["predicted_class"] == "diet_dairy"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
with open("../results/deployment_for_permutations/meat_microbial_top_10.txt", "w") as f:
    for feature in meat_micro:
        f.write(feature + "\n")
with open("../results/deployment_for_permutations/dairy_microbial_top_10.txt", "w") as f:
    for feature in dairy_micro:
        f.write(feature + "\n")
with open("../results/deployment_for_permutations/meat_metabolites_top_10.txt", "w") as f:
    for feature in meat_metab:
        f.write(feature + "\n")
with open("../results/deployment_for_permutations/dairy_metabolites_top_10.txt", "w") as f:
    for feature in dairy_metab:
        f.write(feature + "\n")
```

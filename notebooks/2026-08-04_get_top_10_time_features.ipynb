```python
import pandas as pd
with open("../data/nodes/microbes.txt", "r") as f:
    microbes = [line.strip() for line in f.readlines()]

with open("../data/nodes/metabolites.txt", "r") as f:
    metabolites = [line.strip() for line in f.readlines()]
```

```python
time_perm = pd.read_csv("../results/permutations/time_significant_mean_confidence.tsv", sep="\t")
time_micro = time_perm[time_perm["feature"].isin(microbes)]
time_metab = time_perm[time_perm["feature"].isin(metabolites)]
time_micro, time_metab
endpoint_metab = time_metab[time_metab["direction"] == "endpoint"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()
baseline_metab = time_metab[time_metab["direction"] == "baseline"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()

with open("../results/permutations/endpoint_metabolites_top_10.txt", "w") as f:
    for feature in endpoint_metab:
        f.write(feature + "\n")
with open("../results/permutations/baseline_metabolites_top_10.txt", "w") as f:
    for feature in baseline_metab:
        f.write(feature + "\n")
    
endpoint_micro = time_micro[time_micro["direction"] == "endpoint"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()
baseline_micro = time_micro[time_micro["direction"] == "baseline"].sort_values(by="p_value", ascending=True).head(10)["feature"].tolist()
with open("../results/permutations/endpoint_microbial_top_10.txt", "w") as f:
    for feature in endpoint_micro:
        f.write(feature + "\n")
with open("../results/permutations/baseline_microbial_top_10.txt", "w") as f:
    for feature in baseline_micro:
        f.write(feature + "\n")
```

```python
time_majority = pd.read_csv("../results/deployment_for_permutations/majority_hits_7.tsv", sep="\t")
time_micro = time_majority[time_majority["feature"].isin(microbes)]
time_metab = time_majority[time_majority["feature"].isin(metabolites)]
endpoint_micro = time_micro[time_micro["predicted_class"] == "time_endpoint"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
baseline_micro = time_micro[time_micro["predicted_class"] == "time_baseline"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
endpoint_metab = time_metab[time_metab["predicted_class"] == "time_endpoint"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
baseline_metab = time_metab[time_metab["predicted_class"] == "time_baseline"].sort_values(by="avg_probability", ascending=False).head(10)["feature"].tolist()
with open("../results/deployment_for_permutations/endpoint_microbial_top_10.txt", "w") as f:
    for feature in endpoint_micro:
        f.write(feature + "\n")
with open("../results/deployment_for_permutations/baseline_microbial_top_10.txt", "w") as f:
    for feature in baseline_micro:
        f.write(feature + "\n")
with open("../results/deployment_for_permutations/endpoint_metabolites_top_10.txt", "w") as f:
    for feature in endpoint_metab:
        f.write(feature + "\n")
with open("../results/deployment_for_permutations/baseline_metabolites_top_10.txt", "w") as f:
    for feature in baseline_metab:
        f.write(feature + "\n")
```
